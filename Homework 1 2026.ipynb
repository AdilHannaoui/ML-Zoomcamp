## Sum of weights

```python
asiatic_cars = df[df["origin"]=="Asia"][["vehicle_weight", "model_year"]]
```

```python
asiatic_cars.head(7)
```

```python
X = asiatic_cars.head(7).to_numpy()
```

```python
import numpy as np
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(X.T).dot(y)
```

```python
w.sum()
```
